# Import the IEEE-CIS Fraud Detection data

Install the repository dependencies first (`pip install -r requirements.txt`). In Colab, install `kagglehub` and `pandas` if needed. Accept the competition rules in Kaggle before downloading.

For a local checkout, optionally copy `.env.example` to `.env` and fill in your Kaggle credentials. Existing environment variables take precedence; `~/.kaggle/kaggle.json` is also supported. Otherwise, use the Kaggle login prompt. Never save credentials in notebook output.


```python
import os
import sys
from pathlib import Path

import kagglehub

repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent

# A local checkout can use the authentication helper from the data-read branch.
# Standalone Colab sessions can authenticate with the login prompt below.
if (repo_root / "kaggle_auth.py").is_file():
    if str(repo_root) not in sys.path:
        sys.path.insert(0, str(repo_root))
    from kaggle_auth import get_kaggle_credentials

    try:
        credentials = get_kaggle_credentials()
    except FileNotFoundError:
        credentials = None
    if credentials:
        os.environ["KAGGLE_USERNAME"] = credentials["username"]
        os.environ["KAGGLE_KEY"] = credentials["key"]
        del credentials

kagglehub.login()
```

```python
path = Path(kagglehub.competition_download("ieee-fraud-detection"))
print("Path to competition files:", path)
print("Available files:", sorted(p.name for p in path.glob("*.csv")))
```

```python
import pandas as pd

sample_submission = pd.read_csv(path / "sample_submission.csv")
test_identity = pd.read_csv(path / "test_identity.csv")
test_transaction = pd.read_csv(path / "test_transaction.csv")
train_identity = pd.read_csv(path / "train_identity.csv")
train_transaction = pd.read_csv(path / "train_transaction.csv")
```
